```python
# 1. User Login Check 
# Stored username and password
username = "admin"
password = "1234"

# Taking input from user
input_username = input("Enter username: ")
input_password = input("Enter password: ")

# Checking login
if input_username == username and input_password == password:
    print("Login Successful")
else:
    print("Invalid Credentials")
```

```python
# 2. Pass / Fail Analyzer
# List of student marks
marks = [45, 78, 90, 33, 60]

pass_count = 0
fail_count = 0

# Check each student's marks
for mark in marks:
    if mark >= 50:
        pass_count = pass_count + 1
    else:
        fail_count = fail_count + 1

# Print final result
print("Total Students:", len(marks))
print("Total Passed Students:", pass_count)
print("Total Failed Students:", fail_count)
```

```python
# 3. Simple Data Cleaner
# Original list of names
names = [" Alice ", "bob", " CHARLIE "]

cleaned_names = []

# Clean each name
for name in names:
    name = name.strip()      # Remove extra spaces
    name = name.lower()      # Convert to lowercase
    cleaned_names.append(name)

# Print cleaned list
print("Cleaned Names:", cleaned_names)
```

```python
# 4. Message Length Analyzer
# List of messages
messages = ["Hi", "Welcome to the platform", "OK"]

# Check each message
for message in messages:
    length = len(message)
    print("Message:", message)
    print("Length:", length)
    
    if length > 10:
        print("Status: Long Message")
    else:
        print("Status: Short Message")
    
    print("----------------------")
```

```python
# 5. Error Message Detector
# List of system logs
logs = ["INFO", "ERROR", "WARNING", "ERROR"]

error_count = 0

# Check each log entry
for log in logs:
    if log == "ERROR":
        error_count = error_count + 1

# Print total error count
print("Total ERROR entries:", error_count)
```
