```python
# Problem 1: Unique Words in a Sentence

# Take input from user
sentence = input("Enter sentence: ")

# Convert sentence into list of words
words = sentence.split()

# Convert list into set to remove duplicates
unique_words = set(words)

# Print unique words
print("Unique Words:", unique_words)

# Print count of unique words
print("Count of Unique Words:", len(unique_words))
```

```python
# Problem 2: Highest Salary from Employee Data

# Employee dictionary
employees = {
    "Ravi": 75000,
    "Anita": 68000,
    "Kiran": 72000
}

# Find employee with highest salary
highest_employee = max(employees, key=employees.get)

# Print result
print("Highest Salary:", highest_employee, "-", employees[highest_employee])
```

```python
# Problem 3: Find Maximum and Minimum Values

# Given list
numbers = [45, 22, 89, 10, 66]

# Initialize max and min with first element
maximum = numbers[0]
minimum = numbers[0]

# Loop through the list
for num in numbers:
    if num > maximum:
        maximum = num
    if num < minimum:
        minimum = num

# Print results
print("List:", numbers)
print("Max:", maximum)
print("Min:", minimum)
```

```python
# Problem 4: Count Products Above a Price Threshold

# Given list of product prices
prices = [450, 1200, 899, 1500, 300]

# Initialize counter
count = 0

# Loop through prices
for price in prices:
    if price > 1000:
        count += 1

# Print result
print("Products above 1000:", count)
```

```python
# Problem 5: Calculate Attendance Percentage

# Given attendance list
attendance = ["P", "P", "A", "P", "P"]

# Count total days
total_days = len(attendance)

# Count number of present days
present_days = 0

for day in attendance:
    if day == "P":
        present_days += 1

# Calculate percentage
percentage = (present_days / total_days) * 100

# Print result
print("Attendance Percentage:", percentage)
```

```python
# Problem 6: Remove Duplicate Phone Numbers

# Given list of phone numbers
phone_numbers = [9876543210, 9123456789, 9876543210]

# Convert list to set to remove duplicates
unique_numbers = set(phone_numbers)

# Print unique phone numbers
print("Unique phone numbers:", unique_numbers)
```

```python
# Problem 7: Count Character Frequency

# Given string
text = "pythonp"

# Create empty dictionary
char_count = {}

# Loop through each character
for ch in text:
    if ch in char_count:
        char_count[ch] += 1
    else:
        char_count[ch] = 1

# Print result
print(char_count)
```

```python
# Problem 8: Convert List to Tuple

# Given list
numbers = [10, 20, 30]

# Convert list to tuple
numbers_tuple = tuple(numbers)

# Print result
print("Tuple:", numbers_tuple)
```

```python
# Problem 9: Check if a Key Exists in a Dictionary

# Given dictionary
employees = {
    "Minal": 75000,
    "Vaishu": 68000,
    "Harshada": 72000
}

# Key to check
key_to_check = "Minal"

# Check if key exists
if key_to_check in employees:
    print("Employee exists")
else:
    print("Employee does not exist")
```

```python
# Problem 10: Calculate Average Marks

# Given list of marks
marks = [70, 75, 80, 65]

# Calculate total marks
total = sum(marks)

# Calculate number of subjects
count = len(marks)

# Calculate average
average = total / count

# Print result
print("Average Marks:", average)
```
