```python
# Problem 1: Social Media – Post Engagement Analyzer

# Function to analyze post engagement
def analyze_engagement(likes_list):
    
    # Step 1: Calculate total likes using loop
    total_likes = 0
    for likes in likes_list:
        total_likes += likes

    # Step 2: Check status
    if total_likes >= 1000:
        status = "Viral Post"
    else:
        status = "Normal Engagement"

    # Step 3: Print result
    print("Total Likes:", total_likes)
    print("Post Status:", status)


# Example data (likes from different platforms or hours)
likes_data = [200, 300, 150, 400]

# Function call
analyze_engagement(likes_data)
```

```python
# Problem 3: Agriculture – Rainfall Adequacy Checker

# Function to check rainfall adequacy
def rainfall_checker(rainfall_data, required_level):

    # Step 1: Calculate total rainfall using loop
    total = 0
    for rain in rainfall_data:
        total += rain

    # Step 2: Calculate average rainfall
    average = total / len(rainfall_data)

    # Step 3: Check adequacy
    if average >= required_level:
        status = "Adequate Rainfall"
    else:
        status = "Inadequate Rainfall"

    # Step 4: Display result
    print("Average Rainfall:", int(average))
    print("Rainfall Status:", status)


# Example data
rainfall = [60, 75, 80, 70, 75]
required = 70

# Function call
rainfall_checker(rainfall, required)
```

```python
# Problem 4: Social Media – Duplicate Account Detection

# Function to detect duplicate usernames
def detect_duplicates(usernames):

    # Convert list to set
    unique_users = set(usernames)

    # Compare lengths
    if len(usernames) != len(unique_users):
        print("Duplicate Accounts Found: Yes")
    else:
        print("Duplicate Accounts Found: No")


# Example usernames list
users = ["ravi123", "anita99", "kiran", "ravi123"]

# Function call
detect_duplicates(users)
```

```python
# Problem 5: Healthcare – Appointment Eligibility Checker

# Function to check appointment permission
def check_permission(years):
    if years >= 18:
        return "Eligible"
    else:
        return "Not Eligible"

# Taking input from user
person_age = int(input("Enter Patient Age: "))

# Calling function
result = check_permission(person_age)

# Displaying result
print("Patient Age:", person_age)
print("Eligibility Status:", result)
```

```python
# Problem 6: Agriculture – Premium Crop Price Filter

# Function to filter premium crop prices
def get_premium_prices(rate_list):
    premium_list = []

    for value in rate_list:
        if value > 2000:
            premium_list.append(value)

    return premium_list


# Sample crop prices list
crop_rates = [1500, 2500, 1800, 3200, 1900]

# Calling function
premium_crops = get_premium_prices(crop_rates)

# Output
print("Premium Crops:", premium_crops)
```

```python
# Problem 7: System Monitoring – Application Health Checker
# Function to check system health
def check_system_health(error_total):
    
    # Condition to evaluate system status
    if error_total == 0:
        status = "Healthy"
    elif error_total <= 5:
        status = "Minor Issues"
    else:
        status = "Critical Issues"
    
    return status


# Sample error count
errors = 7

# Function call
result = check_system_health(errors)

# Output
print("Error Count:", errors)
print("System Status:", result)
```

```python
#Problem 8: Banking – Daily Transaction Limit Checker
# Function to check transaction limit
def check_transaction(amount):
    
    daily_limit = 50000   # Daily transaction limit
    
    if amount <= daily_limit:
        status = "Approved"
    else:
        status = "Rejected"
    
    return status


# Sample transaction amount
transaction_amount = 60000

# Function call
result = check_transaction(transaction_amount)

# Output
print("Transaction Amount:", transaction_amount)
print("Transaction Status:", result)
```

```python
# Problem 9: E-Learning – Student Attendance Eligibility System
# Function to calculate attendance percentage and eligibility
def check_attendance(attendance_list):

    total_days = len(attendance_list)
    present_days = 0

    # Loop to count present days
    for day in attendance_list:
        if day == 1:     # 1 means present
            present_days += 1

    # Calculate percentage
    percentage = (present_days / total_days) * 100

    # Check eligibility
    if percentage >= 75:
        status = "Eligible"
    else:
        status = "Not Eligible"

    return percentage, status


# Sample attendance list (1 = Present, 0 = Absent)
attendance = [1,1,1,0,1,1,1,1,0,1]

# Function call
percent, result = check_attendance(attendance)

# Output
print("Attendance Percentage:", percent)
print("Exam Eligibility:", result)
```

```python
# Problem 10: Smart Electricity Bill Analyzer

# Function to calculate electricity bill and usage level
def calculate_bill(units):

    # Bill calculation using slab rates
    if units <= 100:
        bill = units * 3

    elif units <= 200:
        bill = (100 * 3) + ((units - 100) * 5)

    else:
        bill = (100 * 3) + (100 * 5) + ((units - 200) * 7)

    # Usage classification
    if bill < 500:
        status = "Low Usage"
    elif bill <= 1500:
        status = "Moderate Usage"
    else:
        status = "High Usage"

    return bill, status


# Example input
units_consumed = 250

# Function call
total_bill, usage = calculate_bill(units_consumed)

# Output
print("Units Consumed:", units_consumed)
print("Total Bill: ₹", total_bill)
print("Usage Level:", usage)
```
